```python
def iphone_plan(iphone_price,mothly_income,installment_months):
  if installment_months is None:
        if mothly_income > iphone_price:
            return 1, iphone_price, "Affordable"
        else:
            return 1, iphone_price, "Insufficient"
  else:
    price_per_month = iphone_price / installment_months
    if mothly_income > price_per_month:
            return installment_months, price_per_month, "Affordable"
    else:
            return installment_months, price_per_month, "Insufficient"

print(iphone_plan(45000,50000,3))

```

```python
employee_all = [
    [15000, 35000, 3000, 0],
    [20000, 50000, 200000],
    [15000, 12000, 16000, 14000]
]

for employee_j in employee_all:
  employee_duty_j = 0
  for salary_i in employee_j:
    if salary_i > 15000:
      duty_month = salary_i * 3 / 100
      employee_duty_j += duty_month
  print(employee_duty_j)
```
