## 1. What Is a Function?

A **function** is a reusable block of code. You give it a name, call it when you need it, and optionally send data into it.

Basic syntax:

```python
def function_name():
    code_to_run
```

````python
# A simple function with no parameter
def say_hello():
    print("Hello, Python!")

say_hello()
````

## 2. Parameters and Arguments

A **parameter** is a variable inside the function definition.
An **argument** is the real value you pass into the function when you call it.

````python
def greet_student(name):
    print("Hello, " + name + "!")

greet_student("Anna")
greet_student("Ben")
````

## 8. Function Inside Function

A function can call another function. This helps you reuse small functions to build a bigger task.

Example idea:
- `add_tax()` calculates tax.
- `checkout_total()` calls `add_tax()` and then creates the final total.

Still no loops in this section.

````python
def add_tax(price, tax_rate):
    return price + (price * tax_rate)


def make_receipt(product_name, price):
    total = add_tax(price, 0.07)
    return product_name + " total: " + str(total)

print(make_receipt("Notebook", 100))
print(make_receipt("Pen", 20))
````

### Exercise 5.1: Calculate Progressive Personal Income Tax from Net Income

Write a function named `calculate_personal_tax(net_income)`.

The function receives **net income** and calculates tax using the Thai Revenue Department progressive tax brackets below:

| Net income bracket | Tax rate |
|---|---:|
| 0 - 150,000 baht | Exempt, 0% |
| 150,001 - 300,000 baht | 5% |
| 300,001 - 500,000 baht | 10% |
| 500,001 - 750,000 baht | 15% |
| 750,001 - 1,000,000 baht | 20% |
| 1,000,001 - 2,000,000 baht | 25% |
| 2,000,001 - 5,000,000 baht | 30% |
| Over 5,000,000 baht | 35% |

Use progressive calculation. This means each part of income is taxed only at its own bracket rate.

````python
# Exercise 5.1
# Create calculate_personal_tax(net_income).
# Hint: use if, elif, and else. Do not use loops.

def calculate_personal_tax(net_income):
    # TODO: replace pass with your code
    if 0 <= net_income <= 150000:
      return 0

    elif 150000 < net_income <= 300000:
      return (net_income - 150000) * 5/100

    elif 300000 < net_income <= 500000:
      return 150000 * 5/100 + (net_income - 300000) * 10/100

    elif 500000 < net_income <= 750000:
      return 150000 * 5/100 + 200000 * 10/100 + (net_income - 500000) * 15/100

    elif 750000 < net_income <= 1000000:
      return 150000 * 5/100 + 200000 * 10/100 + 250000 * 15/100 + (net_income - 750000) * 20/100

    elif 1000000 < net_income <= 2000000:
      return 150000 * 5/100 + 200000 * 10/100 + 250000 * 15/100 + 250000 * 20/100 + (net_income - 1000000) * 25/100

    elif 2000000 < net_income <= 5000000:
      return 150000 * 5/100 + 200000 * 10/100 + 250000 * 15/100 + 250000 * 20/100 + 1000000 * 25/100 + (net_income - 2000000) * 30/100

    elif net_income > 5000000:
      return 150000 * 5/100 + 200000 * 10/100 + 250000 * 15/100 + 250000 * 20/100 + 1000000 * 25/100 + 3000000 * 30/100 + (net_income - 5000000) * 35/100

print(calculate_personal_tax(100000))   # expected: 0
print(calculate_personal_tax(300000))   # expected: 7500.0
print(calculate_personal_tax(500000))   # expected: 27500.0
print(calculate_personal_tax(1000000))  # expected: 115000.0
print(calculate_personal_tax(6000000))  # expected: 1615000.0
````

### Exercise 5.2: Break-Even Point in Units

Write a function named `calculate_break_even(fixed_cost, selling_price, variable_cost_per_unit)`.

The function receives:
- total fixed cost
- selling price per unit
- variable cost per unit

It returns the number of units that must be sold to break even.

Formula:

$$	ext{Break-Even Units} = rac{	ext{Fixed Cost}}{	ext{Price} - 	ext{Variable Cost}}$$

Assume `selling_price` is greater than `variable_cost_per_unit`.

````python
# Exercise 5.2
# Create calculate_break_even(fixed_cost, selling_price, variable_cost_per_unit).
# Hint: contribution margin = selling_price - variable_cost_per_unit

def calculate_break_even(fixed_cost, selling_price, variable_cost_per_unit):
    # TODO: replace pass with your code
    return fixed_cost / (selling_price - variable_cost_per_unit)

print(calculate_break_even(100000, 250, 150))  # expected: 1000.0
print(calculate_break_even(45000, 120, 75))    # expected: 1000.0
print(calculate_break_even(30000, 100, 40))    # expected: 500.0
````
