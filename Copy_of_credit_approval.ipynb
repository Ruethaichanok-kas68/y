```python
credit_score = int(input("Enter your Credit Score: "))
income_month = int(input("Enter your Monthly Income: l  "))
if credit_score >= 700 and income_month >= 50000:
  print ("Approve")
elif credit_score < 600 or income_month < 20000:
  print ("Reject")
else:
  print ("Wait for Review")
```
